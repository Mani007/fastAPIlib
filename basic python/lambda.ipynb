```python
lc = [x for x in range(1000) if x<=200] # list comprehension for range 200
print(lc)
```

```python
lc[198:]
```

```python
# map
# Even number from first 200 number
even = list(map(lambda x:  x%2 == 0  , lc))
print(even)
```

```python
even = list(filter(lambda x:  x%2 == 0  , lc))
print(even)
```

```python
# reduce function
from functools import reduce
cumulative = reduce(lambda x,y: x+y,lc)
print(cumulative)
```

```python
# reduce function 
from functools import reduce
new_red = reduce(lambda x, y: x if x>y else y , lc) # return the highest of list
print(new_red)
```
