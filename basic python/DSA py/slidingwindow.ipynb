```python
# Sliding window DSA
# 3. Leetcode
# longest substring without repeating a character from a given string
# We need to use sets here
def longest_substring(s:str)->int:
    l = 0
    sett = set()
    n = len(s)
    longest_res = 0
    for r in range(n):
        while s[r] in sett:
            sett.remove(s[l])
            l += 1
        window_size = (r-l)+1
        longest_res = max(longest_res,window_size)
        sett.add(s[r])
    return longest_res

# test
my_str = 'abbcaabb'
print(longest_substring(my_str))

```

```python
# 643 Leet code 
# Fixed sliding window questions
# maximum subarray with size k
def max_avg(nums: list[int], k:int)->float:
    n = len(nums)
    curr_sum = 0
    max_avg_res = 0
    for i in range(k):
        curr_sum += nums[i]
    max_avg_res = curr_sum/k
    for i in range(k,n):
        curr_sum += nums[i]
        curr_sum -= nums[i-k]
        avg = curr_sum/k
        max_avg_res = max(max_avg_res,avg)
    return max_avg_res

testdata = [2,3,4,5,6,7,8,9]
k = 3
print(max_avg(testdata,k))
        
```
