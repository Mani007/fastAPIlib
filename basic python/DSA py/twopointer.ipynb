```python
# Leetcode 977:square of a sorted array
# two pointer methods
def two_pointer(nums:list[int] )-> list[int]: # python with type annotation
    left = 0
    right = len(nums) - 1
    results = []
    while left <= right:
        if abs(nums[left]) > abs(nums[right]):
            results.append(nums[left]**2)
            left += 1
        else:
            results.append(nums[right]**2)
            right -= 1
    results.reverse()
    return results
n = [-3,-7,0,4,6]
res = two_pointer(n)
print(res) # the result is perfect as we are sorting given array based on their square
```
