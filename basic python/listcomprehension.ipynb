```python
# List comprehension with loop
list_with_loop = [x for x in range(10)]
print(list_with_loop)
```

```python
# List comprehension with loop
list_with_loop = [x+1 for x in range(10)]
print(list_with_loop)
```

```python
# Loop version for the same problem
list_loop = []
for x in range(10):
    list_loop.append(x)
print(list_loop)
```

```python
# List comprehension with loop and conditional
list_with_loop = [x for x in range(10) if x>=5]
print(list_with_loop)
```

```python
# Loop version for the above
list_loop1 = []
for x in range(10):
    if x>=5:
        list_loop1.append(x)
print(list_loop1)
```

```python
# List comprehension with loop and conditional
list_with_loop = [x for x in range(20) if x%2 == 0] # list of all the even number
print(list_with_loop)
```

```python
# List comprehension chaining multiple if statement
list_chain = [
    x 
    for x in range(50)
    if x%2 == 0
    if x%3 == 0
    ]
print(list_chain) # printing multiple of 2 and 3
```

```python
# nested list comprehension 
# start with simple for loop
twod = [[1,2,3],[2,3,4],[3,4,5]] # 2D array
flatten_twod = []
for row in twod:
    for col in row:
        flatten_twod.append(col)        
print(flatten_twod) # printing all the numbers
```

```python
# nested list comprehension 
twod = [[1,2,3],[2,3,4],[3,4,5]] # 2D array
flatten_twod = [col for row in twod for col in row] # nested list comprehension done
print(flatten_twod)
```

```python
# categories
category = [ 'even' if x%2 == 0 else 'Odd' for x in range(10) ] # categorizing inside the list
print(category)
```

```python
# creating nested list 
#threed = [[[x for x in range(3)], y for y in range(3) ], z for z in range(3)]
#threed = [[[x for x in range(3)] y for y in range(3) ] z for z in range(3)]
threed = [[[x for x in range(3)] for _ in range(3) ] for _ in range(3)] # _ anonimous variable
print(threed)
```

```python
# applying function to list comprehension
def square(num):
    return num**2
sq_list = [square(x) for x in range(10)]
print(sq_list) 
```

```python
# Dictionary comprehension
somepairs = [('key1',10),('key2',20),('key3',30)]
dictversion = {k:v for k,v in somepairs} # converting list of tuple into a dictionary
print(dictversion)
```

```python
# sets comprehension
nums = [1,1,2,2,2,3,3,4,4,4,5]
uniquesqr = {x**2 for x in nums} # if we do not pass key  in dict then its a set
print(uniquesqr)
```
