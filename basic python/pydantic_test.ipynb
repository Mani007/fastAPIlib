```python
from pydantic import BaseModel
class User(BaseModel):
    id:int
    name:str
    is_active:bool
    
input_data = {"id":'101',"name":"ABC","is_active":True}
user1 = User(**input_data) # We need to provide this ** unpacking dictionary to key value pairs, * only unpack keys
user2 = User(id= input_data['id'],name= input_data['name'],is_active= input_data['is_active']) # We need to provide this ** unpacking dictionary to key value pairs, * only unpack keys
print(user2)
```

```python
from pydantic import BaseModel
from typing import List, Dict, Optional
class Student(BaseModel):
    user_id:int
    user_name:str = "ABC"
    subject:List[str]
    passed:Optional[bool]
    
stu1_data = {
    "user_id":12,
    "user_name":"JJ",
    "subject":["A","B"],
    "passed":True
}
stu1 = Student(**stu1_data)
print(stu1)
```

```python
from pydantic import BaseModel
from typing import List, Dict, Optional
class Student(BaseModel):
    user_id:int
    user_name:str = "ABC"
    subject:List[str]
    passed:Optional[bool]
    
stu1_data = {
    "user_id":12,
    "user_name":"JJ",
    "subject":["A","B"],
    "passed":True
}
stu1 = Student(**stu1_data)
class User(BaseModel):
    student:Student   

user1 = User(student=stu1)
print(user1)
```
