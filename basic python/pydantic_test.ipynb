```python
from pydantic import BaseModel
class User(BaseModel):
    id:int
    name:str
    is_active:bool
    
input_data = {"id":'101',"name":"ABC","is_active":True}
user1 = User(**input_data) # We need to provide this ** unpacking dictionary to key value pairs, * only unpack keys
user2 = User(input_data.values()) # We need to provide this ** unpacking dictionary to key value pairs, * only unpack keys
print(user2)
```
